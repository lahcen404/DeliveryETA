# Modeling

In this notebook, we train machine learning models to predict delivery time.

We use the preprocessed training data to train the models and the test data to evaluate their performance.


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

## Load Preprocessed Data

We load the training and test datasets created during preprocessing.


```python
X_train = pd.read_csv("../data/X_train_preprocessed.csv")
X_test = pd.read_csv("../data/X_test_preprocessed.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze() # squeeze() converts the one-column DataFrame into a Series
y_test = pd.read_csv("../data/y_test.csv").squeeze()
```

## Verify Loaded Data

We check the shapes of the loaded datasets before training the model.


```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## Linear Regression

Linear Regression is a supervised learning algorithm used to predict a numerical value.

We use it as our first baseline model to predict `Time_taken(min)`.


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

## Train the Model

We train the Linear Regression model using the training features and their corresponding delivery times.


```python
model.fit(
    X_train,
    y_train
)

print("Model trained successfully.")
```

```python
print(model)
```

## Linear Regression — Evaluation

After training the Linear Regression model, we evaluate its performance on the test set using several regression metrics:

* **MAE**: average prediction error in minutes.
* **MSE**: average squared prediction error.
* **RMSE**: prediction error in minutes, with more importance given to large errors.
* **R²**: proportion of the variation in delivery time explained by the model.
* **Adjusted R²**: R² adjusted according to the number of features used by the model.

These metrics allow us to evaluate the model from different perspectives and compare it with other regression models.


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predictions
y_pred = model.predict(X_test)

# Metrics
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

# Adjusted R²
n = len(y_test)       # number of observations
p = X_test.shape[1]   # number of features

adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)

# Display results
print("Linear Regression Results")
print("-------------------------")
print("MAE        :", round(mae, 2), "minutes")
print("MSE        :", round(mse, 2))
print("RMSE       :", round(rmse, 2), "minutes")
print("R²         :", round(r2, 4))
print("Adjusted R²:", round(adjusted_r2, 4))
```

## Decision Tree Regression

Decision Tree Regression predicts a numerical value by splitting the data into smaller groups using feature values.

We use it as a second model to compare its performance with Linear Regression.


```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor(
    random_state=42
)
```

## Train the Decision Tree

We train the Decision Tree using the training features and the corresponding delivery times.


```python
tree_model.fit(
    X_train,
    y_train
)

print("Decision Tree trained successfully.")
```

## Make Predictions

We use the trained Decision Tree to predict delivery times for the test data.


```python
tree_pred = tree_model.predict(X_test)

print("First 10 predictions:")
print(tree_pred[:10])
```

## Evaluate the Decision Tree

We evaluate the Decision Tree using MAE, MSE, RMSE, and R².

These metrics allow us to compare the Decision Tree with our Linear Regression baseline.


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predictions
tree_pred = tree_model.predict(X_test)

# Metrics
tree_mae = mean_absolute_error(y_test, tree_pred)
tree_mse = mean_squared_error(y_test, tree_pred)
tree_rmse = np.sqrt(tree_mse)
tree_r2 = r2_score(y_test, tree_pred)

# Adjusted R²
n = len(y_test)
p = X_test.shape[1]

tree_adjusted_r2 = 1 - (1 - tree_r2) * (n - 1) / (n - p - 1)

# Display results
print("Decision Tree Results")
print("---------------------")
print("MAE        :", round(tree_mae, 2), "minutes")
print("MSE        :", round(tree_mse, 2))
print("RMSE       :", round(tree_rmse, 2), "minutes")
print("R²         :", round(tree_r2, 4))
print("Adjusted R²:", round(tree_adjusted_r2, 4))
```

## Compare the Baseline Models

The Decision Tree and Linear Regression models have similar performance on the test set.

The Decision Tree slightly reduces MAE, while Linear Regression has slightly better MSE, RMSE, and R².

These results show that neither model has a clear advantage at this stage.


## Random Forest Regression

Random Forest is an ensemble regression algorithm that combines multiple Decision Trees.

Using several trees can improve generalization compared with a single Decision Tree and can help reduce overfitting.


```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

## Train the Random Forest

We train the Random Forest using the training data.

The test data remains unseen during training.


```python
rf_model.fit(
    X_train,
    y_train
)

print("Random Forest trained successfully.")
```

## Make Predictions

We use the trained Random Forest to predict delivery times for the test data.


```python
rf_pred = rf_model.predict(X_test)

print("First 10 predictions:")
print(rf_pred[:10])
```

## Evaluate the Random Forest

We evaluate the Random Forest using MAE, MSE, RMSE, and R².

These metrics will later be compared with the other candidate models.


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predictions
rf_pred = rf_model.predict(X_test)

# Metrics
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_mse = mean_squared_error(y_test, rf_pred)
rf_rmse = np.sqrt(rf_mse)
rf_r2 = r2_score(y_test, rf_pred)

# Adjusted R²
n = len(y_test)
p = X_test.shape[1]

rf_adjusted_r2 = 1 - (1 - rf_r2) * (n - 1) / (n - p - 1)

# Display results
print("Random Forest Results")
print("---------------------")
print("MAE        :", round(rf_mae, 2), "minutes")
print("MSE        :", round(rf_mse, 2))
print("RMSE       :", round(rf_rmse, 2), "minutes")
print("R²         :", round(rf_r2, 4))
print("Adjusted R²:", round(rf_adjusted_r2, 4))
```

### Interpretation

The Random Forest model achieves a **MAE of 3.67 minutes**, which means that the model's predictions are wrong by approximately **3.67 minutes on average**.

The model also achieves an **RMSE of 4.72 minutes** and an **R² of 0.7496**, meaning that it explains approximately **74.96% of the variation in delivery time** on the test set.

Compared with Linear Regression and a single Decision Tree, Random Forest provides better test-set performance. It is therefore a strong candidate for the next step: **hyperparameter optimization**.


## Train vs Test Evaluation — Overfitting

A model should perform well not only on the training data, but also on unseen test data.

We compare the training and test performance using **MAE, RMSE, R², and Adjusted R²**.

If the training performance is much better than the test performance, the model may be **overfitting**.

A small difference between training and test performance indicates better generalization.


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np


def evaluate_model(model, X_train, y_train, X_test, y_test):
    # Predictions
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    # ---------- TRAIN ----------
    train_mae = mean_absolute_error(y_train, train_pred)
    train_mse = mean_squared_error(y_train, train_pred)
    train_rmse = np.sqrt(train_mse)
    train_r2 = r2_score(y_train, train_pred)

    n_train = len(y_train)
    p_train = X_train.shape[1]

    train_adjusted_r2 = (
        1 - (1 - train_r2) * (n_train - 1)
        / (n_train - p_train - 1)
    )

    # ---------- TEST ----------
    test_mae = mean_absolute_error(y_test, test_pred)
    test_mse = mean_squared_error(y_test, test_pred)
    test_rmse = np.sqrt(test_mse)
    test_r2 = r2_score(y_test, test_pred)

    n_test = len(y_test)
    p_test = X_test.shape[1]

    test_adjusted_r2 = (
        1 - (1 - test_r2) * (n_test - 1)
        / (n_test - p_test - 1)
    )

    return {
        "Train MAE": train_mae,
        "Test MAE": test_mae,
        "Train MSE": train_mse,
        "Test MSE": test_mse,
        "Train RMSE": train_rmse,
        "Test RMSE": test_rmse,
        "Train R²": train_r2,
        "Test R²": test_r2,
        "Train Adjusted R²": train_adjusted_r2,
        "Test Adjusted R²": test_adjusted_r2
    }


# Evaluate all three models
linear_results = evaluate_model(
    model,
    X_train,
    y_train,
    X_test,
    y_test
)

tree_results = evaluate_model(
    tree_model,
    X_train,
    y_train,
    X_test,
    y_test
)

rf_results = evaluate_model(
    rf_model,
    X_train,
    y_train,
    X_test,
    y_test
)


# Display results
for model_name, results in [
    ("Linear Regression", linear_results),
    ("Decision Tree", tree_results),
    ("Random Forest", rf_results)
]:
    print(f"\n{model_name}")
    print("-" * len(model_name))

    print("Train MAE        :", round(results["Train MAE"], 2), "minutes")
    print("Test MAE         :", round(results["Test MAE"], 2), "minutes")

    print("Train MSE        :", round(results["Train MSE"], 2))
    print("Test MSE         :", round(results["Test MSE"], 2))

    print("Train RMSE       :", round(results["Train RMSE"], 2), "minutes")
    print("Test RMSE        :", round(results["Test RMSE"], 2), "minutes")

    print("Train R²         :", round(results["Train R²"], 4))
    print("Test R²          :", round(results["Test R²"], 4))

    print("Train Adjusted R²:", round(results["Train Adjusted R²"], 4))
    print("Test Adjusted R² :", round(results["Test Adjusted R²"], 4))
```

## Overfitting Analysis

The comparison between training and test performance shows different behaviors for the three models.

**Linear Regression** has very similar training and test performance, indicating no significant overfitting. However, its predictive performance remains limited.

**Decision Tree** achieves perfect performance on the training data (`R² = 1.00`) but much lower performance on the test data (`R² = 0.5355`). This indicates strong overfitting: the tree has learned the training data too specifically.

**Random Forest** also shows a gap between training and test performance (`R² = 0.9647` vs `0.7496`), indicating some overfitting. However, its test performance is substantially better than the single Decision Tree, with a **MAE of 3.67 minutes**, **RMSE of 4.72 minutes**, and **R² of 0.7496**.

Random Forest is therefore selected as the candidate model for the next step: **hyperparameter optimization**.


## Hyperparameter Optimization — Random Forest

The Random Forest model achieved the best performance among the initial models, but the difference between training and test performance indicates some overfitting.

To improve the model's generalization, we optimize its hyperparameters using `RandomizedSearchCV`.

`RandomizedSearchCV` tests a number of randomly selected combinations of hyperparameters and evaluates them using cross-validation on the training data.

The test set is not used during the optimization process. It will remain reserved for the final evaluation of the optimized model.

The main hyperparameters we will optimize are:

* `n_estimators`: number of trees in the forest.
* `max_depth`: maximum depth of each tree.
* `min_samples_split`: minimum number of samples required to split a node.
* `min_samples_leaf`: minimum number of samples required in a leaf.
* `max_features`: number of features considered when searching for the best split.


```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV

rf = RandomForestRegressor(
    random_state=42,
    n_jobs=-1
)

param_distributions = {
    "n_estimators": [100, 150, 200],
    "max_depth": [10, 15, 20, 25],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": [0.7, 1.0]
}

random_search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_distributions,
    n_iter=10,
    cv=3,
    scoring="neg_root_mean_squared_error",
    random_state=42,
    n_jobs=-1,
    verbose=1
)

random_search.fit(X_train, y_train)

print("Best parameters:")
print(random_search.best_params_)

print("\nBest CV RMSE:")
print(round(-random_search.best_score_, 2), "minutes")
```

## Evaluation of the Optimized Random Forest

After hyperparameter optimization, we evaluate the best Random Forest on both the training and test datasets.

We calculate MAE, MSE, RMSE, R², and Adjusted R².

The test results are used to measure the model's performance on unseen data.

The difference between training and test performance is also analyzed to detect possible overfitting.


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Best optimized model
best_rf = random_search.best_estimator_

# Predictions
train_pred = best_rf.predict(X_train)
test_pred = best_rf.predict(X_test)

# =========================
# Training metrics
# =========================

train_mae = mean_absolute_error(y_train, train_pred)
train_mse = mean_squared_error(y_train, train_pred)
train_rmse = np.sqrt(train_mse)
train_r2 = r2_score(y_train, train_pred)

n_train = len(y_train)
p_train = X_train.shape[1]

train_adjusted_r2 = 1 - (
    (1 - train_r2) * (n_train - 1)
    / (n_train - p_train - 1)
)

# =========================
# Test metrics
# =========================

test_mae = mean_absolute_error(y_test, test_pred)
test_mse = mean_squared_error(y_test, test_pred)
test_rmse = np.sqrt(test_mse)
test_r2 = r2_score(y_test, test_pred)

n_test = len(y_test)
p_test = X_test.shape[1]

test_adjusted_r2 = 1 - (
    (1 - test_r2) * (n_test - 1)
    / (n_test - p_test - 1)
)

# =========================
# Results
# =========================

print("Optimized Random Forest")
print("-----------------------")

print("\nTraining:")
print("MAE        :", round(train_mae, 2), "minutes")
print("MSE        :", round(train_mse, 2))
print("RMSE       :", round(train_rmse, 2), "minutes")
print("R²         :", round(train_r2, 4))
print("Adjusted R²:", round(train_adjusted_r2, 4))

print("\nTest:")
print("MAE        :", round(test_mae, 2), "minutes")
print("MSE        :", round(test_mse, 2))
print("RMSE       :", round(test_rmse, 2), "minutes")
print("R²         :", round(test_r2, 4))
print("Adjusted R²:", round(test_adjusted_r2, 4))
```

## Evaluation of the Optimized Random Forest

The optimized Random Forest achieves a test MAE of **3.57 minutes** and a test RMSE of **4.55 minutes**.

In business terms, the model's delivery-time predictions are wrong by approximately **3.57 minutes on average**.

The model achieves a test R² of **0.7669**, meaning that it explains approximately **76.69% of the variation** in delivery time.

The training R² is **0.8458**, compared with **0.7669** on the test set. This difference indicates some overfitting, but the model still generalizes well to unseen data.

Compared with the initial Random Forest, the optimized model improves the test MAE from **3.67 to 3.57 minutes**, the RMSE from **4.72 to 4.55 minutes**, and the R² from **0.7496 to 0.7669**.

The optimized Random Forest is therefore selected as the final candidate model.
