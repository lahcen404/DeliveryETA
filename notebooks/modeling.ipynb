# Modeling

In this notebook, we train machine learning models to predict delivery time.

We use the preprocessed training data to train the models and the test data to evaluate their performance.


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

## Load Preprocessed Data

We load the training and test datasets created during preprocessing.


```python
X_train = pd.read_csv("../data/X_train_preprocessed.csv")
X_test = pd.read_csv("../data/X_test_preprocessed.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze() # squeeze() converts the one-column DataFrame into a Series
y_test = pd.read_csv("../data/y_test.csv").squeeze()
```

## Verify Loaded Data

We check the shapes of the loaded datasets before training the model.


```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## Linear Regression

Linear Regression is a supervised learning algorithm used to predict a numerical value.

We use it as our first baseline model to predict `Time_taken(min)`.


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

## Train the Model

We train the Linear Regression model using the training features and their corresponding delivery times.


```python
model.fit(
    X_train,
    y_train
)

print("Model trained successfully.")
```

```python
print(model)
```

```python
print("Intercept:", model.intercept_)
print("Number of coefficients:", len(model.coef_))
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:10])
```

```python
print("Predicted:", y_pred[:10])
print("Actual:   ", y_test.iloc[:10].values)
```

## Evaluate with MAE

MAE measures the average absolute difference between the predicted delivery time and the actual delivery time.


```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

## Evaluate with MSE

MSE (Mean Squared Error) calculates the average squared difference between the predicted and actual delivery times.

Because the errors are squared, large errors have a greater impact on the MSE.


```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)

print("MSE:", round(mse, 2))
```

## Evaluate with RMSE

RMSE (Root Mean Squared Error) is the square root of MSE.

It measures the prediction error in the same unit as the target, which is minutes in our project.

A lower RMSE indicates better prediction performance.


```python
import numpy as np

rmse = np.sqrt(mse)

print("RMSE:", round(rmse, 2), "minutes")
```

## Evaluate with R²

R² (R-squared) measures how well the model explains the variation in the target variable.

Its value is usually between 0 and 1 for a useful regression model.

A value closer to 1 means that the model explains more of the variation in the target.


```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R²:", round(r2, 4))
```
